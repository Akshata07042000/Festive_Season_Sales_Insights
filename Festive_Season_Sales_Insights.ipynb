# 1. Reading data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
df = pd.read_csv('Festive_Sales_Data.csv', encoding = 'unicode_escape')
```

```python
df.shape
```

```python
df.head(10)
```

```python
df.info()
```

# 2. Data cleaning 

```python
# deleting 2 columns 
df.drop(['Status', 'unnamed1'], axis=1 , inplace= True)
```

```python
df.head()
```

```python
pd.isnull(df)
```

```python
# checking null values 
pd.isnull(df).sum()
```

```python
# deleting null values 

df.dropna(inplace = True)
```

```python
pd.isnull(df).sum()
```

```python
# change datatype of Amount Column
df['Amount'] =df['Amount'].astype('int')
```

```python
df['Amount'].dtype
```

```python
df.columns
```

```python
# Rename Column name

df.rename(columns = {'Cust_name' : 'Customer_name'}, inplace='true')
```

```python
df.head(2)
```

```python
df[['Age','Orders','Amount']].describe()
```

# 3 Exploratory Data Analysis

```python
ax = sns.countplot(x='Gender', data=df, hue='Gender')
for bars in ax.containers:
    ax.bar_label(bars)
```

* Above graph shows that most of the buyers are female

```python
df.groupby(['Gender'], as_index= False)['Amount'].sum().sort_values(by="Amount", ascending=False)
```

```python
Sales_gen = df.groupby(['Gender'], as_index= False)['Amount'].sum().sort_values(by="Amount", ascending=False)
ax = sns.barplot(x= 'Gender' , y='Amount' , data= Sales_gen , hue='Gender')

for bars in ax.containers:
   ax.bar_label(bars)
```

* From Above graph the purchasing power of females are greater than mens.

```python
ax = sns.countplot(x='Age Group', data= df , hue= 'Gender')
for bars in ax.containers:
    ax.bar_label(bars)
```

* From above graph we can see that most of the buyers are of 26-35 years female 

```python
# total amount vs Age group 

sales_age = df.groupby(['Age Group'], as_index= False)['Amount'].sum().sort_values(by='Amount', ascending = False)
sns.barplot(x = 'Age Group',y = 'Amount',  data = sales_age, hue='Age Group')
```

* From above graph we can see that most of the buyers who spent more are of age 26-35

```python
# Total number of amount/ sales from 10 states

sales_state=df.groupby(['State'],as_index=False)['Amount'].sum().sort_values(by= 'Amount', ascending=False).head(15)
sns.set(rc={'figure.figsize':(23,5)})
sns.barplot(x='State' , y = 'Amount' , data = sales_state, hue='State')

```

* By above graph we can see that the most of the amount spent are from uttar pradesh , Maharashtra, Karnataka 

```python
# Total number of orders from 10 states

sales_state=df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by= 'Orders', ascending=False).head(15)
sns.set(rc={'figure.figsize':(23,5)})
sns.barplot(x='State' , y = 'Orders' , data = sales_state, hue='State')
```

* By above graph we can see that the most of the orders are from UP , Maharastra and Karnataka  

```python
# Total number of order based on marital status
ax = sns.countplot(x='Marital_Status', data= df)
sns.set(rc={'figure.figsize':(5,5)})
for bars in ax.containers:
    ax.bar_label(bars)

```

* By above graph we can see that most of the orders are placed by married people 

```python
# Total number of Amount based on marital status
sales_status= df.groupby(['Marital_Status', 'Gender'], as_index=False)['Amount'].sum().sort_values(by= 'Amount', ascending=False)
sns.set(rc={'figure.figsize':(6,5)})
sns.barplot(x='Marital_Status' , y = 'Amount' , data = sales_status, hue= 'Gender')
```

* By above graph we can see that most of the amount spent are by married people

```python
ax = sns.countplot(x='Occupation', data= df, hue='Occupation')
sns.set(rc={'figure.figsize':(30,5)})
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
* By above graph we can see oders are placed 
```

```python
Sales_Occupation = df.groupby(['Occupation'], as_index= False)['Amount'].sum().sort_values(by="Amount", ascending=False)
ax = sns.barplot(x='Occupation' , y='Amount' , data= Sales_Occupation , hue='Occupation')

for bars in ax.containers:
   ax.bar_label(bars)
```

* From above graph we can see that most of the buyers are working in IT, HealthCare, Aviation

```python
# Product category
```

```python
ax = sns.countplot(x='Product_Category', data= df, hue='Product_Category')
sns.set(rc={'figure.figsize':(30,5)})
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
ProductCategory = df.groupby(['Product_Category'], as_index= False)['Amount'].sum().sort_values(by="Amount", ascending=False).head(10)
ax = sns.barplot(x='Product_Category' , y='Amount' , data= ProductCategory  , hue='Product_Category')
sns.set(rc={'figure.figsize':(30,5)})
for bars in ax.containers:
   ax.bar_label(bars)
```

* From above graphs we can see that most of the sold products are from Food , Clothingr and electronics category 

```python
Product_ID = df.groupby(['Product_ID'], as_index= False)['Amount'].sum().sort_values(by="Amount", ascending=False).head(10)
ax = sns.barplot(x='Product_ID' , y='Amount' , data= Product_ID ,hue= 'Product_ID')
sns.set(rc={'figure.figsize':(30,5)})
for bars in ax.containers:
   ax.bar_label(bars)
```
